```python
import matplotlib.pyplot as plt
import numpy as np

dias = ["Seg", "Ter", "Qua", "Qui", "Sex", "Sab", "Dom"]
temperatura_media = [24.5, 25.1, 23.8, 26.3, 27.0, 28.2, 26.8]
```

```python
# Find the hottest day
max_temp_index = np.argmax(temperatura_media)
max_temp_day = dias[max_temp_index]
max_temp_value = temperatura_media[max_temp_index]

plt.figure(figsize=(12, 7)) # Slightly larger figure for professionalism
plt.plot(dias, temperatura_media, marker='o', linestyle='-', color='skyblue', linewidth=2, markersize=8, label='Temperatura Média') # Professional line style and color

# Highlight the hottest day
plt.scatter(max_temp_day, max_temp_value, color='red', s=200, zorder=5, label=f'Dia Mais Quente ({max_temp_value}°C)') # Larger, red marker
plt.annotate(f'{max_temp_value}°C',
             (max_temp_day, max_temp_value),
             textcoords="offset points",
             xytext=(0,10),
             ha='center',
             fontsize=12,
             color='red',
             fontweight='bold')

plt.title('Temperatura Média Semanal', fontsize=16, fontweight='bold') # Professional title
plt.xlabel('Dia da Semana', fontsize=14) # Professional x-axis label
plt.ylabel('Temperatura Média (°C)', fontsize=14) # Professional y-axis label
plt.grid(True, linestyle='--', alpha=0.7) # Improved grid
plt.legend(fontsize=12) # Add legend
plt.yticks(np.arange(min(temperatura_media)-1, max(temperatura_media)+2, 1)) # Set y-axis ticks for better readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```
